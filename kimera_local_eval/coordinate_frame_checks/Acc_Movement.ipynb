```python
import pandas as pd
import numpy as np

import seaborn as sns
```

```python
dir_name = 'testing_back_oak_table'

df = pd.read_csv(f'../data/coordinate_frame_checks/{dir_name}/imu0/acc_data.csv')
```

```python
def remove_days(date_str):
    return date_str[7:]
```

```python
df['#timestamp [ns]'] = df['#timestamp [ns]'].apply(remove_days)
```

```python
df['#timestamp [ns]'] = pd.to_timedelta(df['#timestamp [ns]'])
```

```python
df.head(5)
```

```python
df['#timestamp [ns]'] = df['#timestamp [ns]'].dt.total_seconds()
```

```python
sns.set_theme(rc={'figure.figsize':(20,8)})
```

```python
col_to_investigate = 'a_RS_S_z [m s^-2]'
```

```python
sns.lineplot(data=df, x='#timestamp [ns]', y=col_to_investigate)
```
